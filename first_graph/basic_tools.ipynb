```python
from langchain.agents import tool
```

```python
@tool
def add(a: int, b: int) -> int:
    """Add a and b

    Args:
        a (int): first int
        b (int): second int

    Returns:
        int: result
    """
    return a + b
```

```python
@tool
def multiply(a: int, b: int) -> int:
    """multiplies a and b

    Args:
        a (int): first int
        b (int): second int

    Returns:
        int: result
    """
    return a * b
```
